# Task 1: Function computes the end-effector position

```python
def forward_kinematics(l1, l2, theta1, theta2):
  theta1_rad = M.radians(theta1)
  theta2_rad = M.radians(theta2)
  x1 = l1 * M.cos(theta1_rad)
  x2 = x1 + l2 * M.cos(theta1_rad + theta2_rad)
  y1 = l1 * M.sin(theta1_rad)
  y2 = y1 + l2 * M.sin(theta1_rad + theta2_rad)
  return ((x1, y1), (x2, y2))
```

# Task 3: Find all angles combinations that makes the arm stretched in a straigh line

```python
def find_combinations(l1, l2):
  for theta2 in range(0, 181):
    for theta1 in range(0, 181):
      (x1, y1), (x2, y2) = forward_kinematics(l1, l2, theta1, theta2)
      d = M.sqrt(round(x2**2, 10)  + round(y2**2 ,10))
      if d == l1 + l2 or (l2 <= l1 and d == l1 - l2):
        print(f"(theta1: {theta1}, theta2: {theta2})")
```

```python
find_combinations(10, 7)
```
