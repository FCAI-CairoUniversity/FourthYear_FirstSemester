```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_california_housing
from scipy import stats

plt.style.use('dark_background')
np.random.seed(42)

california = fetch_california_housing()
X = np.column_stack([california.data, california.target])
```

```python
X
```

```python
feature_names = list(california.feature_names) + ['HousePrice']
feature_names
```

```python
a = X[0, :2]
b = X[5, :2]
diff = b - a

a
```

```python
fig, ax = plt.subplots(figsize= (8, 6))

ax.quiver(0, 0, a[0], a[1], angles='xy', scale_units='xy', scale=1, color='cyan', label = 'a')
ax.quiver(0, 0, b[0], b[1], angles='xy', scale_units='xy', scale=1, color='lime', label = 'b')

ax.quiver(0, 0, diff[0], diff[1], angles='xy', scale_units='xy', scale=1, color='red', label = 'b - a' )
ax.set_xlim(-10, 10)
ax.set_ylim(0, 60)

ax.set_ylabel('Y')
ax.set_xlabel('X')
plt.show()
```

```python
def dot_product(u, v):
  return np.sum(u * v, axis=1)

def euclidian_distance(u, v):
  return np.sqrt(np.sum((u - v) ** 2, axis=1))

def vector_norm(u):
  return np.sqrt(np.sum(u ** 2))

def cosine_similarity(u, point):
  return dot_product(u, point) / (vector_norm(u) * vector_norm(point))

samples = X[:1000]
point = X[100]

cosine = cosine_similarity(samples, point)
euclidian = euclidian_distance(samples, point)
print(f'Max cosine similarity: {np.sort(cosine)[-2]}')
print(f'Min Euclidian: {np.sort(euclidian)[1]}')
```
